# Student Performance Data Analysis

This notebook contains data analysis for a sample student performance dataset.

## Objectives
1. Load and explore the dataset
2. Perform basic statistical analysis
3. Visualize key findings
4. Draw conclusions about student performance

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set display options
pd.set_option('display.max_columns', None)
plt.style.use('ggplot')
```

## 1. Data Loading and Initial Exploration

```python
# Load the dataset
df = pd.read_csv('../data/sample_data.csv')

# Display basic information
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nFirst 5 rows:")
df.head()
```

```python
# Check data types and missing values
print("Data Types:")
print(df.dtypes)
print("\nMissing Values:")
print(df.isnull().sum())
```

## 2. Descriptive Statistics

```python
# Numerical summary statistics
df.describe()
```

```python
# Course distribution
print("Course Distribution:")
print(df['course'].value_counts())
print("\nGender Distribution:")
print(df['gender'].value_counts())
```

## 3. Data Visualization

```python
# Create figure with subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# GPA distribution
axes[0, 0].hist(df['gpa'], bins=10, edgecolor='black', color='steelblue')
axes[0, 0].set_title('GPA Distribution')
axes[0, 0].set_xlabel('GPA')
axes[0, 0].set_ylabel('Frequency')

# Exam scores by course
df.boxplot(column='exam_score', by='course', ax=axes[0, 1])
axes[0, 1].set_title('Exam Scores by Course')
axes[0, 1].set_xlabel('Course')
axes[0, 1].set_ylabel('Exam Score')
plt.suptitle('')

# Study hours vs Exam score
axes[1, 0].scatter(df['study_hours'], df['exam_score'], c='green', alpha=0.6)
axes[1, 0].set_title('Study Hours vs Exam Score')
axes[1, 0].set_xlabel('Study Hours')
axes[1, 0].set_ylabel('Exam Score')

# Attendance vs GPA
axes[1, 1].scatter(df['attendance'], df['gpa'], c='purple', alpha=0.6)
axes[1, 1].set_title('Attendance vs GPA')
axes[1, 1].set_xlabel('Attendance (%)')
axes[1, 1].set_ylabel('GPA')

plt.tight_layout()
plt.savefig('../reports/analysis_plots.png', dpi=150)
plt.show()
```

## 4. Correlation Analysis

```python
# Calculate correlation matrix
numeric_cols = ['age', 'gpa', 'attendance', 'study_hours', 'exam_score']
correlation_matrix = df[numeric_cols].corr()

# Display correlation matrix
print("Correlation Matrix:")
print(correlation_matrix.round(3))

# Create heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.savefig('../reports/correlation_heatmap.png', dpi=150)
plt.show()
```

## 5. Key Findings and Conclusions

### Summary:
1. **Study Hours Impact**: Students who study more hours tend to have higher exam scores
2. **Attendance Matters**: Higher attendance is positively correlated with better GPA
3. **GPA and Exam Performance**: Strong positive correlation between GPA and exam scores
4. **Course Distribution**: Relatively balanced distribution across courses

### Recommendations:
- Encourage students to increase study hours for better exam performance
- Implement attendance tracking and incentive programs
- Provide additional support for students with lower GPA

```python
# Summary statistics by course
summary_by_course = df.groupby('course')[['gpa', 'attendance', 'study_hours', 'exam_score']].mean()
print("Average Performance by Course:")
summary_by_course.round(2)
```
